```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
from scipy import stats

sns.set_style('whitegrid')

csv_path = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv'

df = pd.read_csv(csv_path)
print('Loaded from:', csv_path)
df.head()
```

---
# PHẦN A — DATA PROFILING
## A.1. Data size, column names, data types

```python
print("Data shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
df.info()
```

## A.2. Missing values & Duplicate data

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicates:", df.duplicated().sum())
df = df.drop_duplicates()
```

## A.3. Invalid values

```python
print(df.describe())
print("\nUnique values of categorical features:")
for col in ['pclass', 'sex', 'embarked', 'who', 'alone']:
    print(col, df[col].unique())
```

## A.4. Create a new column
Tạo cột `family_size` = sibsp + parch + 1.

```python
df['family_size'] = df['sibsp'] + df['parch'] + 1
df.head()
```

# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
print("Mean Age & Fare:", df[['age', 'fare']].mean())
print("Median Age & Fare:", df[['age', 'fare']].median())
print("Mode Age & Fare:\n", df[['age', 'fare']].mode().iloc[0])
```

## Group 2 — Dispersion

```python
print("Range:")
print(df[['age', 'fare']].max() - df[['age', 'fare']].min())
print("\nIQR:")
print(df[['age', 'fare']].quantile(0.75) - df[['age', 'fare']].quantile(0.25))
print("\nStandard Deviation:")
print(df[['age', 'fare']].std())
```

## Group 3 — Location and Shape

```python
print("Percentiles (25%, 50%, 75%):")
print(df[['age', 'fare']].quantile([0.25, 0.50, 0.75]))
print("\nSkewness:\n", df[['age', 'fare']].skew())
print("Kurtosis:\n", df[['age', 'fare']].kurtosis())
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
pclass_survival = df.groupby('pclass')['survived'].mean()
diff = pclass_survival.max() - pclass_survival.min()
print(pclass_survival)
print(f"Chênh lệch giữa hạng cao nhất và thấp nhất: {diff:.4f} ({diff*100:.2f}%)")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
print("Survival rate by Sex:")
print(df.groupby('sex')['survived'].mean())

print("\nSurvival rate by Pclass:")
print(df.groupby('pclass')['survived'].mean())

print("\nSurvival rate by Sex and Pclass:")
print(df.groupby(['sex', 'pclass'])['survived'].mean().unstack())
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
median_fare = df.pivot_table(index='survived', values='fare', aggfunc='median')
print("Median fare by survival status:")
print(median_fare)
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
family_stats = df.pivot_table(
    index='family_size', 
    values='survived', 
    aggfunc=['count', 'mean']
)

family_stats.columns = ['Count', 'Survival_Rate']
print(family_stats)
```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
embark_survival = df.pivot_table(
    index='embark_town', 
    values='survived', 
    aggfunc=['count', 'mean']
)

embark_survival.columns = ['Count', 'Survival_Rate']
print(embark_survival)
```
